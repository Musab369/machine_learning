```python
#creates array for normal array
arr= list(range(1000000))

#Normal python list
start = time.time()
result = [x*2 for x in arr]
end = time.time()
print("Normal python time;",end-start)
```

```python
#creates array for numpy array
p_arr = np.array(arr)

#Numpy array
start= time.time()
result= p_arr*2
end= time.time()
print("Numpy Time:",end-start)
```
