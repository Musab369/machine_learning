```python
# how to measure execution time:
import time
arr = np.array([4,5])
start = time.time()
print('np.sum() =>',np.sum(arr))
end = time.time()
s = end-start
print('Execution time:',s)
# np.sum() is adds of all elements of one array
```

```python
arr = np.array([4,5])
z = arr.cumsum(axis = 0)
start = time.time()
print(z[1])
end = time.time()
o = end-start
print('Execution time:',o)
```

```python
arr = np.array([4,5])
z = np.add.reduce(arr,axis=0)
start = time.time()
print(z)
end = time.time()
t = end-start
print('Execution time:',t)

# So we can use:
 # 1. np.cumsum(arr,axis)
 # 2. np.add.reduce(arr,axis)
# Because these are faster than np.sum() to add elements of same array..
```

```python
if (o>s and o>t):
    print('o lose!')
elif o<s and o<t:
    print('o won!')
elif t>s and t>o:
    print('t won!')
elif t<s and t<o:
    print('t won!')
else:
    print('s won!')
```

```python
# In Cartesian coordinates, a point is described using two numbers: (x,y)
# x = horizontal position (left/right)
# y = vertical position (up/down)

# In polar coordinates, a point is described using: (r,θ)
# r = distance from the origin (center)
# θ (theta) = angle from the positive x-axis (usually measured in degrees or radians)

# np.sin()
# np.cos()
# np.tan()
# np.arcsin() → inverse of sin
# np.arccos() → inverse of cos
# np.arctan() → inverse of tan

np.random.seed(42)
arr = np.random.randint(1,10,(10,2))
print('-----------Cartesian Coordinates--------\n',arr)


r = np.sqrt(np.sum(np.power(arr,2),axis=1))
r = np.round(r,2).reshape(10,1)
print('r:\n',r)

# theta = np.arctan(np.divide(arr))
# np.round(theta,2)

theta = np.arctan(np.divide(arr[0:,1],arr[0:,0]))
theta = np.degrees(theta)
theta = np.round(theta,2).reshape(10,1)
print('theta:\n',theta)

# arr[0:,0]
# arr[0:,1]
print('-----------Polar Coordinates--------\n(r,0)=\n',np.concat((r,theta),axis=1))

```
