```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ali", "John", "Sara"]
})

df2 = pd.DataFrame({
    "id": [2, 3, 4],
    "marks": [90, 78, 85]
})
```

```python
# Inner merge: Keeps only common ids.
pd.merge(df1,df2, on="id", how="inner")
```

```python
# Outer merge: Keeps everything from both.
pd.merge(df1,df2, on="id", how="outer")
```

```python
# Right merge: Keeps all rows from left (df2).
pd.merge(df1,df2, on="id", how="right")
```

```python
# Left merge: Keeps all rows from left (df1).
pd.merge(df1,df2, on="id", how="left")
```

```python
# groupby()
import pandas as pd

Data = {
    "Name":['A','B','C','D','E'],
    "Department":['HR','IT','HR','IT','HR'],
    "Salary":[4000,5000,4500,5500,3500]
}
df = pd.DataFrame(Data)
df
```

```python
# Sum:
df.groupby("Department")["Salary"].sum()
```

```python
# Group by one column:
df.groupby("Department")
```

```python
# Mean or average salary per department:
df.groupby("Department")["Salary"].mean()
```

```python
# Count:
df.groupby("Department")["Salary"].count()
```

```python
# Max/Min:
max = df.groupby("Department")["Salary"].max()
min = df.groupby("Department")["Salary"].min()
print(max)
print(min)
```

```python
# Multiple Aggregations:
df.groupby(["Department"])["Salary"].agg(["sum","min","mean","max","count"])
```

```python
# Group by Multiple Columns:
df.groupby(["Department","Name"])["Salary"].count()
```

```python
# Pivot-Table:
data = {
    "Name":['A','B','A','D','B','D'],
    "Product":["Pen","Book","Book","Book","Pen","Pen"],
    "Sale":[100,200,300,150,400,500]
}
df = pd.DataFrame(data)
df
```

```python
df.pivot_table(
    values = "Sale",
    index = "Name",
    columns = "Product",
    aggfunc = "sum"
)
```
