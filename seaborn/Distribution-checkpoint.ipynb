```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='survived',hue='sex')
plt.title('No of Male/Female survived after titanic incident!')
plt.grid()
```

```python
sns.displot(data=titanic,x='age',hue='sex',kind='hist')
plt.title('No of Male/Female had travelled on titanic on that day!')
plt.grid()
```
