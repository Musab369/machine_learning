```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tiped = sns.load_dataset('tips')
tiped
```

```python
sns.scatterplot(data=tiped, x='total_bill', y='tip'
                ,hue='sex', style = 'time')
plt.title('Hotel Manager Report')
plt.grid()
plt.show()
```

```python
sns.relplot(data=tiped, x='total_bill', y='tip'
                ,kind = 'scatter',hue='sex', style = 'time')
plt.title('Hotel Manager Report')
plt.grid()
plt.show()
```
