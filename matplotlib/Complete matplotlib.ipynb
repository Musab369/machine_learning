```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
price = [48000,54000,57000,49000,47000,40000]
year = [2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
plt.grid()
```

```python
batsman = pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color="blue", marker="o", label="Kohli")
plt.plot(batsman['index'],batsman['RG Sharma'],color="black", marker="o", label="Rohit")
plt.grid()
plt.title('Rohit Sharma Vs Virat Kohli Score Comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
plt.show()
```

```python
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,label='2015',color='blue')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,label='2016',color='red')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,label='2017',color='yellow')

plt.title('Batsman Season Record in IPL between 2015-2017')
plt.xlabel('Batsman')
plt.ylabel('Runs scored')
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.legend()
plt.show()


```
